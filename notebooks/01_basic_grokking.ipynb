# 01: Basic Grokking Phenomenon

This notebook demonstrates the basic grokking phenomenon on the modular arithmetic task.
It trains a small transformer model on the `(a+b) mod 59` dataset in the `pure` condition (no data collapse) and visualizes the characteristic delay between training accuracy convergence and test accuracy convergence (the "grokking gap").

```python
!pip install torch numpy matplotlib
import sys
import os
# Ensure the source directory is accessible
if '..' not in sys.path:
    sys.path.append('..')
```

```python
from src.train import TrainConfig, train

# Run training for a short duration to demonstrate the effect
config = TrainConfig(
    condition_name="pure_demo",
    collapse_level=0.0,
    max_steps=10000,
    eval_every=100,
    output_dir="results_notebook/01"
)

state = train(config)
```

```python
import matplotlib.pyplot as plt

steps = [e['step'] for e in state.history]
train_acc = [e['train_acc'] for e in state.history]
test_acc = [e['test_acc'] for e in state.history]

plt.figure(figsize=(10, 5))
plt.plot(steps, train_acc, label='Train Acc')
plt.plot(steps, test_acc, label='Test Acc')
plt.axvline(x=state.grokking_step, color='r', linestyle='--', label='Grokking Step')
plt.title('Basic Grokking: (a+b) mod 59')
plt.xlabel('Step')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```
