# 02: The Effect of Model Collapse on Grokking

This notebook demonstrates how injecting synthetic "collapsed" data into the training set monotonically delays or prevents grokking.
We will run two conditions: `pure` and `severe_collapse` and compare their test accuracy curves.

```python
!pip install torch numpy matplotlib
import sys
import os
if '..' not in sys.path:
    sys.path.append('..')
```

```python
from src.train import TrainConfig, train

print("Training pure condition...")
config_pure = TrainConfig(
    condition_name="pure",
    collapse_level=0.0,
    max_steps=10000,
    eval_every=100,
    output_dir="results_notebook/02"
)
state_pure = train(config_pure)

print("Training severe_collapse condition...")
config_collapse = TrainConfig(
    condition_name="severe_collapse",
    collapse_level=0.5,
    collapse_severity=0.9,
    max_steps=10000,
    eval_every=100,
    output_dir="results_notebook/02"
)
state_collapse = train(config_collapse)
```

```python
import matplotlib.pyplot as plt

steps_pure = [e['step'] for e in state_pure.history]
test_acc_pure = [e['test_acc'] for e in state_pure.history]

steps_collapse = [e['step'] for e in state_collapse.history]
test_acc_collapse = [e['test_acc'] for e in state_collapse.history]

plt.figure(figsize=(10, 5))
plt.plot(steps_pure, test_acc_pure, label='Pure Test Acc')
plt.plot(steps_collapse, test_acc_collapse, label='Severe Collapse Test Acc')
plt.title('Effect of Model Collapse on Grokking')
plt.xlabel('Step')
plt.ylabel('Test Accuracy')
plt.legend()
plt.show()
```
