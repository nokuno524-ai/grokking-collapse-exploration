# 03: Scaling Analysis and Grokking Cliffs

This notebook demonstrates how to load data from a grid of experiments and analyze the "grokking cliffs" - how weight decay and noise fraction interact to create a sharp phase transition in grokking behavior.
We use simulated results here for quick demonstration on Colab.

```python
!pip install numpy matplotlib
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Simulated data based on paper findings
wds = np.array([0.1, 0.3, 1.0, 3.0])
noises = np.array([0.0, 0.05, 0.10, 0.15, 0.20])

# 1 if grokked, 0 otherwise. 
# Findings: wd=3.0 prevents grokking. 
# For wd=0.3 and 1.0, groks at noise <= 0.10, cliff at 0.15.
grok_matrix = np.array([
    [1, 1, 1, 0, 0],  # wd=0.1
    [1, 1, 1, 0, 0],  # wd=0.3
    [1, 1, 1, 0, 0],  # wd=1.0
    [0, 0, 0, 0, 0],  # wd=3.0
])

plt.figure(figsize=(8, 6))
plt.imshow(grok_matrix, cmap='coolwarm', aspect='auto')
plt.colorbar(ticks=[0, 1], label='Grokked (1=Yes, 0=No)')
plt.xticks(np.arange(len(noises)), noises)
plt.yticks(np.arange(len(wds)), wds)
plt.xlabel('Noise Fraction')
plt.ylabel('Weight Decay')
plt.title('Grokking Cliff Phase Diagram')
plt.show()
```
